```python
import pandas as pd

# 1. Load the data
df = pd.read_csv('../data/raw_crypto_prices.csv', index_col='Date', parse_dates=True)

# 2. Check for missing values
print("Missing values per coin:\n", df.isnull().sum())

# 3. Handle missing values 
# For time-series financial data, 'forward fill' (carrying the last known price forward) is standard
df_cleaned = df.ffill().bfill() # bfill handles any NaNs at the very beginning

# 4. Feature Engineering: Create new metrics for analysis
# Calculate the 30-Day Moving Average for Bitcoin
df_cleaned['BTC_30MA'] = df_cleaned['BTC-USD'].rolling(window=30).mean()

# Calculate daily percentage returns
returns = df_cleaned[['BTC-USD', 'ETH-USD', 'SOL-USD']].pct_change().dropna()

# Save cleaned data
df_cleaned.to_csv('../data/cleaned_crypto_prices.csv')
```
